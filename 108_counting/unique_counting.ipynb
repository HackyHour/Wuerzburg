```python
import random


# create a function for rounds
def do_round(storage_length, stream):
    turn = 0
    storage = {}
    lenarray = []
    # so lange Liste nicht voll, schreibe rein
    for word in stream:
            if word in storage:
                # do stuff
                storage[word] += 1
                for toss in range(turn):
                    if random.random() <= 0.5:
                        del storage[word]
                        break
            else:
                keep = True
                for t in range(turn):
                    if (random.random() <= 0.5):
                        keep = False
                if keep:
                    storage[word] = 1
                    
                # check if list full
                while len(storage) >= storage_length:
                    random_erase_half(storage)
                    turn += 1
            lenarray.append(ausrechnen(turn,storage))
    return turn, storage, lenarray
        

# randomly erase half the data
def random_erase_half(storage):
    for key in list(storage.keys()):
        if random.random() <= 0.5:
            del storage[key]

def ausrechnen(turn, storage):
    return len(storage)*2**turn
```

```python
# this will be our storage location
storagelength=128

turn, storage,lenarray = do_round(storagelength, stream)
print(turn)
print(storage)

ausrechnen(turn, storage)
```

```python
import nfc
```

```python
import libusb1
```

```python
libusb1.libusb_init(None)
# list usb devices
print(libusb1.libusb_device_descriptor())
```

```python
clf = nfc.ContactlessFrontend()
```

```python
print(clf)
```

```python
def onconnect(tag):
    print("Something happened")
    print(tag.identifier.hex())
    if hasattr(tag, "dump"):
        print(tag.dump())
    return True
```

```python
clf.connect(rd={"on-connect": onconnect})
```

```python
print(clf.open('usb:001:015'))    # open device 6 on bus 1
```

```python
print(clf.open('usb:001:015'))    # open device 6 on bus 1
print(clf.open('usb:0c4b:9102'))
print(clf.open('usb:001'))        # open first Reader on bus 1
print(clf.open('usb:0c4b'))       # open first Sony Reader
print(clf.open('usb'))  
print(clf.open('tty'))  
```
